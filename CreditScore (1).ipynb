```python
def clean_numeric(col, dtype=float):
    col = (
        col.astype(str)
           .str.replace(r"[^0-9.]", "", regex=True)
           .replace("", np.nan)
    )
    return col.astype(dtype)




#drop ID-like columns
df = df.drop(columns=["Name", "ID", "Customer_ID", "SSN"])

# month → numericdf["Month"] = pd.to_datetime(df["Month"], format="%B").dt.month

df["Age"] = clean_numeric(df["Age"], int)
df["Annual_Income"] = clean_numeric(df["Annual_Income"], float)
df["Num_of_Loan"] = clean_numeric(df["Num_of_Loan"], int)
df["Num_of_Delayed_Payment"] = clean_numeric(df["Num_of_Delayed_Payment"], float)
df["Changed_Credit_Limit"] = clean_numeric(df["Changed_Credit_Limit"], float)
df["Outstanding_Debt"] = clean_numeric(df["Outstanding_Debt"], float)
df["Amount_invested_monthly"] = clean_numeric(df["Amount_invested_monthly"], float)
df["Monthly_Balance"] = clean_numeric(df["Monthly_Balance"], float)

```

```python
num_cols=["Annual_Income","Num_of_Loan","Monthly_Inhand_Salary","Changed_Credit_Limit","Num_of_Delayed_Payment",
          "Num_Credit_Inquiries","Outstanding_Debt","Amount_invested_monthly","Monthly_Balance"]

for col in num_cols:
    df[col] = df[col].fillna(df[col].median())

cat_cols=["Type_of_Loan","Credit_History_Age"]

for col in cat_cols:
    df[col] = df[col].fillna(df[col].mode()[0])
```

```python
df["Age"] = pd.to_numeric(df["Age"], errors="coerce")
median_age = df.loc[df["Age"] > 0, "Age"].median()
df.loc[df["Age"] < 0, "Age"] = median_age
```

```python
df = df[df["Age"].between(0, 100)]
```

```python
df['Age'].describe()
```

```python
plt.figure()
plt.boxplot(df["Age"])
plt.title("Boxplot of Age")
plt.ylabel("Age")
plt.show()
```

```python
Q1 = df["Age"].quantile(0.25)
Q3 = df["Age"].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

df = df[(df["Age"] >= lower) & (df["Age"] <= upper)]
```

```python
plt.figure()
plt.boxplot(df["Age"].dropna())
plt.title("Boxplot of Age")
plt.ylabel("Age")
plt.show()
```

```python
for col in num_cols:
    df[col]=np.log1p(df[col])
    plt.figure()
    plt.boxplot(df[col])
    plt.title(f"Boxplot of {col}")
    plt.ylabel(col)
    plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

cols = num_cols
n = len(cols)
rows = int(np.ceil(n / 3))

fig, axes = plt.subplots(rows, 3, figsize=(15, 4 * rows))
axes = axes.flatten()

for i, col in enumerate(cols):
    df[col + "_log"] = np.log1p(df[col])
    axes[i].boxplot(df[col + "_log"].dropna())
    axes[i].set_title(f"{col} (log)")
    axes[i].set_ylabel(col)

# remove empty plots
for j in range(i + 1, len(axes)):
    fig.delaxes(axes[j])

plt.tight_layout()
plt.show()
```
